**BOOK CLASS**

```python
class Book:
    def __init__(self, book_id, title, author):
        self.book_id = book_id
        self.title = title
        self.author = author
        self._is_available = True

    def borrow(self):
        """Mark the book as borrowed."""
        if not self._is_available:
            raise ValueError(f"Book '{self.title}' is already borrowed.")
        self._is_available = False

    def return_book(self):
        """Mark the book as available again."""
        if self._is_available:
            raise ValueError(f"Book '{self.title}' was not borrowed.")
        self._is_available = True

    def is_available(self):
        """Return True if the book can be borrowed."""
        return self._is_available

    def __str__(self):
        status = "Available" if self._is_available else "Borrowed"
        return f"[{self.book_id}] {self.title} by {self.author} - {status}"

```

**class Member**

```python

class Member:
    """A regular library member (borrow limit = 3)."""

    member_type = "Member"   # shown in __str__

    def __init__(self, member_id, name):
        self.member_id = member_id
        self.name = name
        self._borrowed_books = []   # list of Book objects (protected-style)

    def get_borrow_limit(self):
        """Regular members can borrow 3 books. Subclasses override this."""
        return 3

    def borrow_book(self, book):
        """Borrow a book after checking the limit and availability."""
        if len(self._borrowed_books) >= self.get_borrow_limit():
            raise ValueError(
                f"{self.name} reached the borrow limit ({self.get_borrow_limit()} books)."
            )
        if not book.is_available():
            raise ValueError(f"Book '{book.title}' is already borrowed.")
        book.borrow()                    # change the book status
        self._borrowed_books.append(book)

    def return_book(self, book):
        """Return a book, only if this member really borrowed it."""
        if book not in self._borrowed_books:
            raise ValueError(f"{self.name} did not borrow '{book.title}'.")
        book.return_book()               # change the book status
        self._borrowed_books.remove(book)

    def list_borrowed_books(self):
        """Return a copy of the borrowed books list (so outside code can't change it)."""
        return list(self._borrowed_books)

    def __str__(self):
        return (f"[{self.member_id}] {self.name} ({self.member_type}) - "
                f"borrowed {len(self._borrowed_books)}/{self.get_borrow_limit()}")


class Customer(Member):
    """Customer: borrow limit = 2."""

    member_type = "Customer"

    def get_borrow_limit(self):
        return 2




```

**library class**

```python
class Library:
    def __init__(self, name):
        self.name = name
        self.books = []
        self.members = []

    # ---------- finding by ID (loop over the list) ----------
    def get_book(self, book_id):
        """Return the Book with this ID, or None if not found."""
        for book in self.books:
            if book.book_id == book_id:
                return book
        return None

    def get_member(self, member_id):
        """Return the Member with this ID, or None if not found."""
        for member in self.members:
            if member.member_id == member_id:
                return member
        return None

    # ---------- adding ----------
    def add_book(self, book):
        if self.get_book(book.book_id) is not None:
            raise ValueError(f"Book ID '{book.book_id}' already exists.")
        self.books.append(book)

    def register_member(self, member):
        if self.get_member(member.member_id) is not None:
            raise ValueError(f"Member ID '{member.member_id}' already exists.")
        self.members.append(member)

    # ---------- searching / displaying ----------
    def find_book(self, query):
        """Search by book ID (exact), or title / author (partial, case-insensitive)."""
        query = query.strip().lower()
        results = []
        for book in self.books:
            if (query == str(book.book_id)
                    or query in book.title.lower()
                    or query in book.author.lower()):
                results.append(book)
        return results

    def display_available_books(self):
        available = [b for b in self.books if b.is_available()]
        if not available:
            print("No available books right now.")
        for book in available:
            print(book)

    # ---------- borrowing / returning ----------
    def _get_member_and_book(self, member_id, book_id):
        """Validate both IDs and return the two objects."""
        member = self.get_member(member_id)
        if member is None:
            raise ValueError(f"Member ID '{member_id}' not found.")
        book = self.get_book(book_id)
        if book is None:
            raise ValueError(f"Book ID '{book_id}' not found.")
        return member, book

    def lend_book(self, member_id, book_id):
        member, book = self._get_member_and_book(member_id, book_id)
        member.borrow_book(book)

    def receive_return(self, member_id, book_id):
        member, book = self._get_member_and_book(member_id, book_id)
        member.return_book(book)

```

```python

def read_text(prompt):
    """Read a non-empty string from the user."""
    value = input(prompt).strip()
    if not value:
        raise ValueError("Input cannot be empty.")
    return value


def read_id(prompt):
    """Read an ID: a whole number 0, 1, 2, 3, ... (no upper limit)."""
    text = read_text(prompt)
    if not text.isdigit():          # rejects letters, negatives, decimals
        raise ValueError("ID must be a whole number (0, 1, 2, ...).")
    return int(text)


def preload_data(library):
    """Pre-load 3 books and 2 members when the program starts."""
    library.add_book(Book(0, "Python Crash Course", "Eric Matthes"))
    library.add_book(Book(1, "Clean Code", "Robert Martin"))
    library.add_book(Book(2, "Introduction to Algorithms", "Thomas Cormen"))
    library.register_member(StudentMember(0, "Ahmed"))
    library.register_member(FacultyMember(1, "Dr. Hadeer"))
    library.register_member(Member(2, "Sara"))


def show_menu():
    print("\n===== Library Menu =====")
    print("1. View Available Books")
    print("2. Search Books")
    print("3. Add Book")
    print("4. Register Member")
    print("5. Borrow Book")
    print("6. Return Book")
    print("7. View Member's Borrowed Books")
    print("8. Exit")


def search_books(library):
    results = library.find_book(read_text("Enter book ID, title or author: "))
    if not results:
        print("No books found.")
    for book in results:
        print(book)


def add_book(library):
    book = Book(read_id("Book ID: "), read_text("Title: "), read_text("Author: "))
    library.add_book(book)
    print("Book added successfully.")


def register_member(library):
    member_id = read_id("Member ID: ")
    name = read_text("Name: ")
    print("Member type: 1) Regular  2) Student  3) Faculty")
    choice = input("Choose (1-3): ").strip()
    types = {"1": Member, "2": StudentMember, "3": FacultyMember}
    if choice not in types:
        raise ValueError("Invalid member type.")
    library.register_member(types[choice](member_id, name))
    print("Member registered successfully.")


def borrow_book(library):
    library.lend_book(read_id("Member ID: "), read_id("Book ID: "))
    print("Book borrowed successfully.")


def return_book(library):
    library.receive_return(read_id("Member ID: "), read_id("Book ID: "))
    print("Book returned successfully.")


def view_member_books(library):
    member_id = read_id("Member ID: ")
    member = library.get_member(member_id)
    if member is None:
        raise ValueError(f"Member ID '{member_id}' not found.")
    print(member)
    books = member.list_borrowed_books()
    if not books:
        print("No borrowed books.")
    for book in books:
        print("  -", book)

```
